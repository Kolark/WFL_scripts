```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../../results/benchmarks/wake_benchmark_1000_100_v5.csv")
df
```

```python
for method_name, group in df.groupby('method'):
    # if method_name=="CALC_WAKE_WRF":
    #     continue
    plt.plot(
        group['total_turbines'],
        group['time_calc_wake'],
        label=method_name,
        marker='o'  # Optional: adds a marker to each data point
    )

# 3. Add labels, title, and legend
# plt.yscale('log')
plt.xlabel('total_turbines')
plt.ylabel('time_calc_wake')
plt.title('Time Calculation Wake vs Grid Size by Method')
plt.legend(title='Method')
plt.grid(True)
```

```python
for method_name, group in df.groupby('method'):
    plt.plot(
        group['total_turbines'],
        group['time_get_wind_data'],
        label=method_name,
        marker='o'  # Optional: adds a marker to each data point
    )

# 3. Add labels, title, and legend
plt.xlabel('total_turbines')
plt.ylabel('time_get_wind_data')
plt.title('Time Calculation Wake vs Grid Size by Method')
plt.legend(title='Method')
plt.grid(True)
```

```python
# 2. Calculating variance
variance_df = df.groupby('n_grid')['ideal_total'].var()
# print("Variance of ideal_total per n_grid:")
print(variance_df)
plt.plot(variance_df)
```
